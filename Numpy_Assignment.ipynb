```python
import numpy as np
names = np.array(["Student1", "Student2", "Student3"])
data = np.array([
    [18, 85, 78],
    [20, 92, 88],
    [19, 75, 95]])
```

```python
# 1.Get the shape of the matrix.
print("Shape:", data.shape)
```

```python
#2.Find the average age of students.
average_age = np.mean(data[:, 0])
print("Average Age:", average_age)
```

```python
# 3.Extract the Math marks of all students.
math_marks = data[:, 1]
print("Math Marks:", math_marks)
```

```python
# 4.Find the highest Science mark.
science_marks = data[:, 2]
print("Highest Science Marks:", np.max(science_marks))
```

```python
# 5.Get the details of the student who scored more than 90 in Math.
result = data[:, 1] > 90
print("Student Names:", names[result])
print("Student Details:", data[result])
```

```python
#6.Increase the Math marks of all students by 5.
data[:, 1] = data[:, 1] + 5
print("Updated Data:")
print(data)
```

```python
# 7.Find how many students are younger than 19.
count = np.sum(data[:, 0] < 19)
print("Students younger than 19:", count)
```

```python
# 8.Calculate the average marks in each subject (column-wise mean).
average_marks = np.mean(data[:, 1:3], axis=0)
print("Average Math Marks:", average_marks[0])
print("Average Science Marks:", average_marks[1])
```

```python
# 9.Get the data of students who scored at least 80 in both subjects.
result = (data[:, 1] >= 80) & (data[:, 2] >= 80)
print("Student Names:", names[result])
print("Student Details:")
print(data[result])
```

```python
#10.Replace all Science marks below 75 with 0.
data[data[:, 2] < 75, 2] = 0
print("Updated Data:")
print(data)
```
